```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('/content/Air_Quality.csv')

# Clean string columns by stripping extra spaces and standardizing case
df['Location'] = df['Location'].str.strip().str.lower()
df['Air_Quality_category'] = df['Air_Quality_category'].str.strip().str.lower()

# --- DATA PREPARATION & AGGREGATION ---

# Filter data for just the city location
city_center_data = df[df['Location'] == 'city']

# Group data by location to find average pollution levels
avg_pollution = df.groupby('Location')[['Air_Quality_category', 'PM2.5_Level']].max()

# --- THE 5 REQUIRED CHARTS ---

# Chart 1: Line Chart (Tracking Air Quality Category over time at City Center)
plt.figure(figsize=(8, 4))
plt.plot(city_center_data['Date'], city_center_data['Air_Quality_category'], marker='o', color='red')
plt.title('Chart 1: Air Quality Category Trend at City Center')
plt.xlabel('Date')
plt.ylabel('Air Quality Category')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Ensure dataset is loaded and cleaned
df = pd.read_csv('/content/Air_Quality.csv')
df['Location'] = df['Location'].str.strip().str.lower()

# Group data by location to find average pollution levels
avg_pollution = df.groupby('Location')[['PM2.5_Level']].mean()

# Chart 2: Bar Chart (Comparing average PM 2.5 by location)
plt.figure(figsize=(8, 4))
avg_pollution['PM2.5_Level'].plot(kind='bar', color='orange')
plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5 Level')
plt.xticks(rotation=0)
plt.show()
```

```python
# Chart 3: Scatter Plot (Distribution of PM2.5 Levels over Dates)
plt.figure(figsize=(8, 4))
plt.scatter(df['Date'], df['PM2.5_Level'], color='purple', s=50)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset locally to ensure 'df' is defined
df = pd.read_csv('/content/Air_Quality.csv')

# Chart 4: Histogram (Distribution of all PM2.5 readings)
plt.figure(figsize=(8, 4))
plt.hist(df['PM2.5_Level'], bins=5, color='red', edgecolor='black')
plt.title('Chart 4: Frequency of Different PM2.5 Levels')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.show()
```

```python
# Chart 5: Horizontal Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].max().plot(kind='barh', color='teal')
plt.title('Chart 5: Maximum PM2.5 Recorded per Location')
plt.xlabel('Peak PM2.5 Level')
plt.ylabel('Location')
plt.show()
```

```python
# Chart 5: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].max().plot(kind='bar', color='teal')
plt.title('Chart 5: Maximum PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Peak PM2.5 Level')
plt.xticks(rotation=0)
plt.show()
```
